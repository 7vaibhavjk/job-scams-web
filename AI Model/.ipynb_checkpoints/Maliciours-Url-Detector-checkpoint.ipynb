```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, random_split
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import re
import urllib.parse
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')
```

```python
class URLFeatureExtractor:
    """Extract lexical features from URLs"""
    
    def __init__(self):
        self.tfidf = TfidfVectorizer(max_features=1000, ngram_range=(1, 3), analyzer='char')
        self.scaler = StandardScaler()
        
    def extract_lexical_features(self, url):
        """Extract lexical features from a single URL"""
        features = {}
        
        # Basic URL properties
        features['url_length'] = len(url)
        features['num_dots'] = url.count('.')
        features['num_hyphens'] = url.count('-')
        features['num_underscores'] = url.count('_')
        features['num_slashes'] = url.count('/')
        features['num_questionmarks'] = url.count('?')
        features['num_equal'] = url.count('=')
        features['num_at'] = url.count('@')
        features['num_and'] = url.count('&')
        features['num_exclamation'] = url.count('!')
        features['num_space'] = url.count(' ')
        features['num_tilde'] = url.count('~')
        features['num_comma'] = url.count(',')
        features['num_plus'] = url.count('+')
        features['num_asterisk'] = url.count('*')
        features['num_hash'] = url.count('#')
        features['num_dollar'] = url.count('$')
        features['num_percent'] = url.count('%')
        
        # Domain and subdomain features
        try:
            parsed = urllib.parse.urlparse(url if url.startswith('http') else 'http://' + url)
            domain = parsed.netloc.lower()
            path = parsed.path
            query = parsed.query
            
            features['domain_length'] = len(domain)
            features['path_length'] = len(path)
            features['query_length'] = len(query)
            features['num_subdomains'] = domain.count('.') - 1 if domain.count('.') > 0 else 0
            
            # Check for suspicious patterns
            features['has_ip'] = int(bool(re.search(r'\d+\.\d+\.\d+\.\d+', domain)))
            features['has_port'] = int(':' in domain)
            
        except:
            features['domain_length'] = 0
            features['path_length'] = 0
            features['query_length'] = 0
            features['num_subdomains'] = 0
            features['has_ip'] = 0
            features['has_port'] = 0
        
        # Entropy calculation
        features['url_entropy'] = self._calculate_entropy(url)
        
        # Character ratios
        features['digit_ratio'] = sum(c.isdigit() for c in url) / len(url) if url else 0
        features['letter_ratio'] = sum(c.isalpha() for c in url) / len(url) if url else 0
        
        return features
```

```python
 def _calculate_entropy(self, string):
        """Calculate Shannon entropy of string"""
        if not string:
            return 0
        entropy = 0
        for x in set(string):
            p_x = string.count(x) / len(string)
            if p_x > 0:
                entropy += - p_x * np.log2(p_x)
        return entropy
```

```python
def fit_transform_features(self, urls, labels=None):
        """Extract features from list of URLs and fit scalers"""
        # Extract lexical features
        lexical_features = []
        for url in tqdm(urls, desc="Extracting lexical features"):
            features = self.extract_lexical_features(url)
            lexical_features.append(list(features.values()))
        
        lexical_features = np.array(lexical_features)
        
        # Fit and transform TF-IDF features
        tfidf_features = self.tfidf.fit_transform(urls).toarray()
        
        # Combine features
        all_features = np.hstack([lexical_features, tfidf_features])
        
        # Scale features
        if labels is not None:  # Training mode
            all_features = self.scaler.fit_transform(all_features)
        
        return all_features
```

```python
def transform_features(self, urls):
        """Transform URLs using fitted scalers"""
        # Extract lexical features
        lexical_features = []
        for url in urls:
            features = self.extract_lexical_features(url)
            lexical_features.append(list(features.values()))
        
        lexical_features = np.array(lexical_features)
        
        # Transform TF-IDF features
        tfidf_features = self.tfidf.transform(urls).toarray()
        
        # Combine features
        all_features = np.hstack([lexical_features, tfidf_features])
        
        # Scale features
        all_features = self.scaler.transform(all_features)
        
        return all_features
```

```python
class URLDataset(Dataset):
    """Custom Dataset for URL classification"""
    
    def __init__(self, features, labels):
        self.features = torch.FloatTensor(features)
        self.labels = torch.LongTensor(labels)
    
    def __len__(self):
        return len(self.features)
    
    def __getitem__(self, idx):
        return self.features[idx], self.labels[idx]

class URLClassifier(nn.Module):
    """Neural Network for URL Classification"""
    
    def __init__(self, input_size, hidden_sizes=[512, 256, 128], num_classes=2, dropout_rate=0.3):
        super(URLClassifier, self).__init__()
        
        layers = []
        prev_size = input_size
        
        # Hidden layers
        for hidden_size in hidden_sizes:
            layers.extend([
                nn.Linear(prev_size, hidden_size),
                nn.ReLU(),
                nn.BatchNorm1d(hidden_size),
                nn.Dropout(dropout_rate)
            ])
            prev_size = hidden_size
        
        # Output layer
        layers.append(nn.Linear(prev_size, num_classes))
        
        self.network = nn.Sequential(*layers)
    
    def forward(self, x):
        return self.network(x)
```

```python
class MaliciousURLDetector:
    """Main class for malicious URL detection"""
    
    def __init__(self, model_params=None):
        self.feature_extractor = URLFeatureExtractor()
        self.model = None
        self.label_encoder = LabelEncoder()
        self.device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        self.model_params = model_params or {}
        
    def load_data(self, csv_path):
        """Load data from CSV file"""
        print(f"Loading data from {csv_path}...")
        df = pd.read_csv(csv_path)
        
        # Assume the CSV has columns 'url' and 'type' or similar
        # Adjust column names based on your dataset structure
        if 'url' in df.columns:
            urls = df['url'].tolist()
        else:
            # If first column is URLs
            urls = df.iloc[:, 0].tolist()
        
        if 'type' in df.columns:
            labels = df['type'].tolist()
        elif 'label' in df.columns:
            labels = df['label'].tolist()
        else:
            # If second column is labels
            labels = df.iloc[:, 1].tolist()
        
        print(f"Loaded {len(urls)} URLs")
        return urls, labels
    
    def preprocess_data(self, urls, labels):
        """Preprocess URLs and labels"""
        print("Preprocessing data...")
        
        # Clean URLs
        urls = [str(url).strip() for url in urls if str(url).strip()]
        labels = labels[:len(urls)]
        
        # Encode labels
        labels_encoded = self.label_encoder.fit_transform(labels)
        
        # Extract features
        features = self.feature_extractor.fit_transform_features(urls, labels_encoded)
        
        print(f"Feature shape: {features.shape}")
        print(f"Classes: {self.label_encoder.classes_}")
        
        return features, labels_encoded
    
    def train(self, csv_path, test_size=0.2, batch_size=64, epochs=50, learning_rate=0.001):
        """Train the model"""
        # Load and preprocess data
        urls, labels = self.load_data(csv_path)
        features, labels_encoded = self.preprocess_data(urls, labels)
        
        # Create dataset
        dataset = URLDataset(features, labels_encoded)
        
        # Split dataset
        train_size = int((1 - test_size) * len(dataset))
        test_size = len(dataset) - train_size
        train_dataset, test_dataset = random_split(dataset, [train_size, test_size])
        
        # Create data loaders
        train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
        test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
        
        # Initialize model
        input_size = features.shape[1]
        self.model = URLClassifier(
            input_size=input_size,
            num_classes=len(np.unique(labels_encoded)),
            **self.model_params
        ).to(self.device)
        
        # Loss and optimizer
        criterion = nn.CrossEntropyLoss()
        optimizer = optim.Adam(self.model.parameters(), lr=learning_rate)
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=5)
        
        # Training loop
        print("Starting training...")
        train_losses = []
        
        for epoch in range(epochs):
            self.model.train()
            epoch_loss = 0
            
            for batch_features, batch_labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}"):
                batch_features, batch_labels = batch_features.to(self.device), batch_labels.to(self.device)
                
                optimizer.zero_grad()
                outputs = self.model(batch_features)
                loss = criterion(outputs, batch_labels)
                loss.backward()
                optimizer.step()
                
                epoch_loss += loss.item()
            
            avg_loss = epoch_loss / len(train_loader)
            train_losses.append(avg_loss)
            scheduler.step(avg_loss)
            
            # Validation
            if (epoch + 1) % 10 == 0:
                val_acc = self.evaluate(test_loader)
                print(f"Epoch {epoch+1}/{epochs}, Loss: {avg_loss:.4f}, Val Accuracy: {val_acc:.4f}")
        
        # Final evaluation
        print("\nFinal evaluation:")
        self.detailed_evaluation(test_loader)
        
    def evaluate(self, test_loader):
        """Evaluate model on test data"""
        self.model.eval()
        correct = 0
        total = 0
        
        with torch.no_grad():
            for batch_features, batch_labels in test_loader:
                batch_features, batch_labels = batch_features.to(self.device), batch_labels.to(self.device)
                outputs = self.model(batch_features)
                _, predicted = torch.max(outputs.data, 1)
                total += batch_labels.size(0)
                correct += (predicted == batch_labels).sum().item()
        
        return correct / total
    
    def detailed_evaluation(self, test_loader):
        """Detailed evaluation with classification report"""
        self.model.eval()
        all_preds = []
        all_labels = []
        
        with torch.no_grad():
            for batch_features, batch_labels in test_loader:
                batch_features, batch_labels = batch_features.to(self.device), batch_labels.to(self.device)
                outputs = self.model(batch_features)
                _, predicted = torch.max(outputs.data, 1)
                
                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(batch_labels.cpu().numpy())
        
        # Convert back to original labels
        original_labels = self.label_encoder.inverse_transform(all_labels)
        original_preds = self.label_encoder.inverse_transform(all_preds)
        
        print(f"Accuracy: {accuracy_score(original_labels, original_preds):.4f}")
        print("\nClassification Report:")
        print(classification_report(original_labels, original_preds))
        print("\nConfusion Matrix:")
        print(confusion_matrix(original_labels, original_preds))
    
    def predict(self, urls):
        """Predict if URLs are malicious or benign"""
        if self.model is None:
            raise ValueError("Model not trained yet. Call train() first.")
        
        # Handle single URL
        if isinstance(urls, str):
            urls = [urls]
        
        # Extract features
        features = self.feature_extractor.transform_features(urls)
        features_tensor = torch.FloatTensor(features).to(self.device)
        
        self.model.eval()
        with torch.no_grad():
            outputs = self.model(features_tensor)
            probabilities = torch.softmax(outputs, dim=1)
            _, predicted = torch.max(outputs, 1)
        
        # Convert predictions back to original labels
        predicted_labels = self.label_encoder.inverse_transform(predicted.cpu().numpy())
        confidence_scores = probabilities.max(dim=1)[0].cpu().numpy()
        
        results = []
        for i, url in enumerate(urls):
            results.append({
                'url': url,
                'prediction': predicted_labels[i],
                'confidence': confidence_scores[i],
                'is_safe': predicted_labels[i].lower() in ['benign', 'safe', '0']
            })
        
        return results if len(results) > 1 else results[0]
    
    def save_model(self, path):
        """Save the trained model"""
        if self.model is None:
            raise ValueError("No model to save")
        
        torch.save({
            'model_state_dict': self.model.state_dict(),
            'feature_extractor': self.feature_extractor,
            'label_encoder': self.label_encoder,
            'model_params': self.model_params
        }, path)
        print(f"Model saved to {path}")
    
    def load_model(self, path):
        """Load a trained model"""
        checkpoint = torch.load(path, map_location=self.device)
        
        self.feature_extractor = checkpoint['feature_extractor']
        self.label_encoder = checkpoint['label_encoder']
        self.model_params = checkpoint['model_params']
        
        # Recreate model architecture
        input_size = len(self.feature_extractor.scaler.mean_)  # Get input size from scaler
        self.model = URLClassifier(
            input_size=input_size,
            num_classes=len(self.label_encoder.classes_),
            **self.model_params
        ).to(self.device)
        
        self.model.load_state_dict(checkpoint['model_state_dict'])
        print(f"Model loaded from {path}")
```
